# 08 streaming

Streaming responses token-by-token — what powers a responsive chat UI.

**Setup** — run once per kernel session. Assumes `.env` is configured at the project root (see `../README.md`) and you've run `gcloud auth application-default login`.

```python
import sys, os
sys.path.append(os.path.abspath(os.path.join('..', 'samples')))
```

## Setup & definitions

```python
from common import DEFAULT_MODEL, get_client


def main():
    client = get_client()

    for chunk in client.models.generate_content_stream(
        model=DEFAULT_MODEL,
        contents="Write a 6-line haiku sequence about distributed systems failing gracefully.",
    ):
        print(chunk.text, end="", flush=True)
    print()
```

## Run it

```python
main()
```
