# 01 basic text

Basic text generation, with a system instruction and generation config.

**Setup** — run once per kernel session. Assumes `.env` is configured at the project root (see `../README.md`) and you've run `gcloud auth application-default login`.

```python
import sys, os
sys.path.append(os.path.abspath(os.path.join('..', 'samples')))
```

## Setup & definitions

```python
from common import DEFAULT_MODEL, get_client
from google.genai import types


def main():
    client = get_client()

    response = client.models.generate_content(
        model=DEFAULT_MODEL,
        contents="Give me three risks of relying only on in-house LLM evals.",
        config=types.GenerateContentConfig(
            system_instruction="You are a terse, skeptical AI platform architect.",
            temperature=0.4,
            max_output_tokens=512,
        ),
    )

    print(response.text)
    print("\n--- usage ---")
    print(response.usage_metadata)
```

## Run it

```python
main()
```
